```python
import pandas as pd

from immunodash.client import ChEMBLClient
from immunodash.targets import get_target
from immunodash.activities import get_activities
from immunodash import cleaning
from immunodash import molecules
from immunodash import descriptors
```

```python
client = ChEMBLClient()

jak1 = get_target("JAK1")

activities = client.get_all(
    endpoint="activity",
    response_key="activities",
    params={
        "target_chembl_id": jak1["target_chembl_id"]
    },
    show_progress=True
)

raw_df = pd.DataFrame(activities)
```

```python
clean = cleaning.clean_activities(raw_df)
```

```python
mols = molecules.get_unique_molecules(clean)
```

```python
desc = descriptors.calculate_rdkit_descriptors(mols)
```

```python
fingerprints = descriptors.calculate_morgan_fingerprints(
    mols.head()
)
```

```python
scaffolds = descriptors.calculate_murcko_scaffolds(
    mols.head()
)
```
