## Imports

```python
import requests
import pandas as pd
import json
```

## Request info

```python
url = "https://www.ebi.ac.uk/chembl/api/data/target/search"

params = {
    "q": "JAK1",
    "format": "json"
}

response = requests.get(url, params=params)
```

```python
response.status_code #check
```

## JSON file

```python
data = response.json()
```

```python
first = data["targets"][0]
first.keys()
```

```python
import json

print(json.dumps(first["target_components"], indent=2))
```
